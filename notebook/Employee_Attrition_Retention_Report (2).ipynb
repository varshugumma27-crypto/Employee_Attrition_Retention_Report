```python
import matplotlib.pyplot as plt
df=pd.read_csv("Cleaned_Employee_data.csv")


df["Department_Region"].value_counts().plot(kind="bar")
plt.title("Number of Employee by Department")
plt.xlabel("Department_Region")
plt.ylabel("Number of Employees")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
print(df.columns)
```

```python
avg_salary = df.groupby("Age")["Total_Salary"].mean()
plt.figure(figsize=(8,5))
plt.plot(avg_salary.index, avg_salary.values, marker="o")
plt.title("Average Salary by Age")
plt.xlabel("Age")
plt.ylabel("Average Salary")
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.hist(df["Total_Salary"], bins=15)
plt.title("Salary Distribution")
plt.xlabel("Total Salary")
plt.ylabel("Number of employees")
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.scatter(df["Age"], df["Total_Salary"])
plt.title("Age vs Total Salary")
plt.xlabel("Age")
plt.ylabel("Total Salary")
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(7,7))

df["Status"].value_counts().plot(kind="pie",autopct="%1.1f%%", startangle=90)

plt.title("Employee Status Distribution")
plt.ylabel("")
plt.show()
```

```python
plt.figure(figsize=(8,5))
df["Performance_Score"].value_counts().sort_index().plot(kind="bar")
plt.title("Employee Performance Score ")
plt.xlabel("Performance Score")
plt.ylabel("Number of Employees")
plt.tight_layout()
plt.show()
```
